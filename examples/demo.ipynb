# TIF-Diff 遥感图像解译示例

这个notebook演示如何使用TIF-Diff系统进行遥感图像解译。

```python
# 导入必要的库
import sys
sys.path.append('..')

from sam_rs_core import RemoteSensingConfig, RemoteSensingSAM
from visualization import RemoteSensingVisualizer
import matplotlib.pyplot as plt
%matplotlib inline
```

## 1. 初始化系统

```python
# 创建配置
config = RemoteSensingConfig(
    sam_checkpoint="../weights/sam_vit_h_4b8939.pth",
    device="cuda"  # 如果没有GPU，改为 "cpu"
)

# 初始化模型
print("初始化SAM模型...")
rs_sam = RemoteSensingSAM(config)

# 初始化可视化器
visualizer = RemoteSensingVisualizer()
```

## 2. 处理遥感图像

```python
# 指定图像路径
image_path = "path/to/your/image.tif"  # 替换为您的图像路径

# 处理图像
print("处理遥感图像...")
results = rs_sam.process_remote_sensing_image(image_path)

print(f"检测到 {len(results['masks'])} 个对象")
```

## 3. 可视化结果

```python
# 显示分割结果
visualizer.visualize_results(results, show_indices=True)
```

## 4. 分析统计

```python
# 统计各类别数量
category_counts = {}
for mask in results['masks']:
    cat = mask.get('category', 'unknown')
    category_counts[cat] = category_counts.get(cat, 0) + 1

print("类别分布:")
for cat, count in category_counts.items():
    print(f"  {cat}: {count}")
```
